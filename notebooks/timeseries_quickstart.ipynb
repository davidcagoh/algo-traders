# Time-series lab quickstart
Search and combine the shared crypto, derivatives, macro, traditional-market, on-chain, DeFi, and sentiment series without copying data.

```python
from timeseries_lab import TimeSeriesLab
lab = TimeSeriesLab()
```

```python
lab.search("vix")
```

```python
btc = lab["spot_1d:BTC:close"]
funding = lab["funding_binance:BTC | resample:1D:sum | rolling_zscore:90"]
btc.tail(), funding.tail()
```

```python
panel = lab.frame({
    "btc_return": "spot_1d:BTC:close | log_return",
    "vix_change": "fred:VIXCLS | diff",
    "two_year_change": "fred:DGS2 | diff",
}, start="2020-01-01", freq="1D", fill="ffill", fill_limit=4)
panel.tail()
```

```python
lab.plot({"BTC": "spot_1d:BTC:close", "S&P 500": "traditional_daily:SPY:close"},
         start="2020-01-01", normalize=True, title="BTC and S&P 500")
```

```python
result = lab.regress(
    "spot_1d:BTC:close | log_return",
    {"vix_change": "fred:VIXCLS | diff", "rates_change": "fred:DGS2 | diff"},
    start="2020-01-01", freq="1D", fill="ffill", fill_limit=4,
)
result.summary
```
